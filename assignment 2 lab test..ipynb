```python


#Lab Experiment 1 using python 
#Generate three independent samples randomly and perform One-way ANOVA f_oneway 
#function and calculate p-value. Interpret if the null hypothesis will be rejected or accepted 
#based on the p-values. 


import numpy as np
from scipy.stats import f_oneway

# Generate three random samples
np.random.seed(42)
sample1 = np.random.randint(10, 50, size=10)
sample2 = np.random.randint(10, 50, size=10)
sample3 = np.random.randint(10, 50, size=10)

# Perform One-way ANOVA
f_stat, p_value = f_oneway(sample1, sample2, sample3)

print("F-statistic:", f_stat)
print("p-value:", p_value)

# Interpretation
if p_value < 0.05:
    print("Reject H₀ → Significant difference between group means.")
else:
    print("Fail to reject H₀ → No significant difference between group means.")
```

```python
 
#Lab Experiment 4 using python 
#Perform one way anova for the following 3 groups. Also perform tuckeyhsd test to 
# if the difference between means is significant for some groups at 5% los. 
#Perform Tuckey HSD test to test signifance of difference between means with each group 
#pairs. 
#group a - 85, 86, 88, 75, 78, 94, 98, 79, 71, 80 
#group b - 91, 92, 93, 90, 97, 94, 82, 88, 95, 96 
#group c - 79, 78, 88, 94, 92, 85, 83, 85, 82, 81 

import numpy as np
from scipy.stats import f_oneway

# Groups
group_a = [85, 86, 88, 75, 78, 94, 98, 79, 71, 80]
group_b = [91, 92, 93, 90, 97, 94, 82, 88, 95, 96]
group_c = [79, 78, 88, 94, 92, 85, 83, 85, 82, 81]

# Perform One-way ANOVA
f_stat, p_value = f_oneway(group_a, group_b, group_c)

print("F-statistic:", f_stat)
print("p-value:", p_value)

if p_value < 0.05:
    print("Reject H₀ → At least one group mean differs significantly.")
else:
    print("Fail to reject H₀ → No significant difference between group means.")
```

```python
#Lab Experiment 5 using python 
#Perform 2-Way ANOVA on the shared dataset on two categories A and B and their 
#values: Two_Way_Anova_Dataset. Print the output in terms of 1. Degree of freedom for 
#residues as "df["Err"]", 2. Mean Square Error for residues as "ms["Err"]", 
#3. Anova for rows as Fischer's value "F1", 
#4. Anova for columns as Fischer's value "F2" 
#and interpret your results.

import pandas as pd
import statsmodels.api as sm
from statsmodels.formula.api import ols

# Dataset from lab sheet
data = {
    'CatA': ['A','A','A','A','B','B','B','B','A','A','A','A','C','C','C','C','B','B','B'],
    'CatB': ['R','S','T','S','R','S','R','S','R','U','U','R','S','S','S','T','S','T','S'],
    'Value': [1.0,1.0,1.0,1.0,2.0,2.0,2.0,2.0,1.0,1.0,1.0,1.0,3.0,3.0,3.0,3.0,2.0,2.0,2.0]
}

df = pd.DataFrame(data)

# Two-way ANOVA model
model = ols('Value ~ C(CatA) + C(CatB) + C(CatA):C(CatB)', data=df).fit()
anova_table = sm.stats.anova_lm(model, typ=2)

print(anova_table)
```
