```python
# Q1 .Solve the Example for Independent sample t-test: 
 
    # Example: Given a summary of data of two samples, apply t-test assuming that 
   # population variances are equal, compare sample means at 5% los 
                 #Sample   Sample   Sample 
                  #Size      Mean       Variance 
#Sample 1    13        15.0          87.5 
#Sample 2    11        12.0         39.0 
 
 
 
#Paired sample t-test:

import math
from scipy import stats

# Given data
n1, mean1, var1 = 13, 15.0, 87.5
n2, mean2, var2 = 11, 12.0, 39.0

# Pooled variance
sp2 = ((n1-1)*var1 + (n2-1)*var2) / (n1+n2-2)

# Standard error
se = math.sqrt(sp2 * (1/n1 + 1/n2))

# t-statistic
t_stat = (mean1 - mean2) / se

# Degrees of freedom
df = n1 + n2 - 2

# p-value (two-tailed)
p_value = 2 * (1 - stats.t.cdf(abs(t_stat), df))

print("t-statistic:", t_stat)
print("p-value:", p_value)

```

```python

#Q2. Use Random seed to generate two samples Data1 and Data2, deriving t-statistics and 
#p-value from the data. Interpret your result at 5% los.  
 
#Chi-squared test:

import numpy as np
from scipy import stats

np.random.seed(42)  # reproducibility

# Generate two samples
data1 = np.random.randint(10, 20, size=15)
data2 = data1 + np.random.randint(-3, 3, size=15)

# Paired t-test
t_stat, p_value = stats.ttest_rel(data1, data2)

print("Data1:", data1)
print("Data2:", data2)
print("t-statistic:", t_stat)
print("p-value:", p_value)
```

```python
 
# Q3.Chi-squared test: 
#Lab Experiment 4 using python:  
 
#i. Check the goodness of fit for the following data mapping number of hours a student 
#studies v/s the expected number of hours at 5% los. 
#Observed Data (in hours) 8 6 10 7 8 11 9
#Expected Data  (in hours) 9 8 11 8 10 7 6

from scipy.stats import chisquare

# Observed and expected data
observed_hours = [8, 6, 10, 7, 8, 11, 9]
expected_hours = [9, 8, 11, 8, 10, 7, 6]

# Perform chi-squared test
chi_stat, p_value = chisquare(f_obs=observed_hours, f_exp=expected_hours)

print("Chi-squared Statistic:", chi_stat)
print("p-value:", p_value)

# Interpretation
if p_value < 0.05:
    print("Reject H₀ → Significant difference between observed and expected data.")
else:
    print("Fail to reject H₀ → No significant difference.")
```

```python
# Q.3.-->ii. Check the goodness of fit for the following data mapping number of hours a student 
#studies v/s the expected number of hours at 5% los. 
#Observed Data (no. of accidents) 15 
#12 
#18 
#30 
#Expected Data  (no. of accidents) 16.7 16.7 16.7 16.6 16.6 16.7 
#Summary 
#10 
#15

from scipy.stats import chisquare

# Observed and expected data
observed_accidents = [15, 12, 18, 10, 15, 30]
expected_accidents = [16.7, 16.7, 16.7, 16.6, 16.6, 16.7]

# Perform chi-squared test
chi_stat, p_value = chisquare(f_obs=observed_accidents, f_exp=expected_accidents)

print("Chi-squared Statistic:", chi_stat)
print("p-value:", p_value)

# Interpretation
if p_value < 0.05:
    print("Reject H₀ → Observed accident pattern differs significantly from expected.")
else:
    print("Fail to reject H₀ → Accident pattern fits expected distribution.")
```
