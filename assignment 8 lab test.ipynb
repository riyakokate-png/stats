```python
#LAB Experiment 1 Give python code find mean, median, mode, first quartile, second 
#quartile, third quartile, interquartile range and outliers for following distribution of 
#Scores: 55, 60, 62, 70, 75, 78, 80, 82, 85, 90. use IQR and Z-score method to detect 
#outliers

import numpy as np
import scipy.stats as stats

# Dataset
scores = np.array([55, 60, 62, 70, 75, 78, 80, 82, 85, 90])

# Mean, Median, Mode
mean_val = np.mean(scores)
median_val = np.median(scores)
mode_val = stats.mode(scores, keepdims=True).mode[0]

# Quartiles
Q1 = np.percentile(scores, 25)
Q2 = np.percentile(scores, 50)  # same as median
Q3 = np.percentile(scores, 75)
IQR = Q3 - Q1

# Outlier detection using IQR
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
outliers_iqr = scores[(scores < lower_bound) | (scores > upper_bound)]

# Outlier detection using Z-score
z_scores = (scores - mean_val) / np.std(scores)
outliers_z = scores[np.abs(z_scores) > 3]

# Results
print("Mean:", mean_val)
print("Median:", median_val)
print("Mode:", mode_val)
print("Q1:", Q1)
print("Q2:", Q2)
print("Q3:", Q3)
print("Interquartile Range (IQR):", IQR)
print("Outliers (IQR method):", outliers_iqr)
print("Outliers (Z-score method):", outliers_z)
```

```python
#LAB Experiment 3. Find Inter-quartile range manually for the following dataset 
#data = [5, 8, 15, 26, 10, 18, 3, 12, 6, 14, 11]


import numpy as np

# Dataset
data = [5, 8, 15, 26, 10, 18, 3, 12, 6, 14, 11]
data_sorted = sorted(data)

# Quartiles
Q1 = np.percentile(data_sorted, 25)
Q2 = np.percentile(data_sorted, 50)
Q3 = np.percentile(data_sorted, 75)
IQR = Q3 - Q1

# Outlier detection
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
outliers = [x for x in data_sorted if x < lower_bound or x > upper_bound]

print("Sorted Data:", data_sorted)
print("Q1:", Q1)
print("Q2 (Median):", Q2)
print("Q3:", Q3)
print("Interquartile Range (IQR):", IQR)
print("Outliers:", outliers)
```

```python
#LAB Experiment 4. Generate two random datasets and using seaborn plot box plot and 
#voilen plot for the datasets

import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# 1. Generate two random datasets
np.random.seed(42)  # for reproducibility
data1 = np.random.normal(loc=50, scale=10, size=100)  # mean=50, std=10
data2 = np.random.normal(loc=60, scale=15, size=100)  # mean=60, std=15

# 2. Combine into a DataFrame
df = pd.DataFrame({
    "Dataset1": data1,
    "Dataset2": data2
})

# Melt the DataFrame for seaborn
df_melt = df.melt(var_name="Dataset", value_name="Values")

# 3. Plot Box Plot
plt.figure(figsize=(8,6))
sns.boxplot(x="Dataset", y="Values", data=df_melt, palette="Set2")
plt.title("Box Plot of Two Random Datasets")
plt.show()

# 4. Plot Violin Plot
plt.figure(figsize=(8,6))
sns.violinplot(x="Dataset", y="Values", data=df_melt, palette="Set3")
plt.title("Violin Plot of Two Random Datasets")
plt.show()
```

```python







import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import statsmodels.api as sm
from statsmodels.stats.outliers_influence import variance_inflation_factor

# 1. Load dataset
df = pd.read_csv("vif_data.csv")

# 2. Calculate VIF for each predictor
X = df.drop(columns=['Target'])   # assuming 'Target' is dependent variable
X_const = sm.add_constant(X)      # add intercept
vif = pd.DataFrame()
vif["Feature"] = X_const.columns
vif["VIF"] = [variance_inflation_factor(X_const.values, i) for i in range(X_const.shape[1])]

print("Variance Inflation Factors:\n", vif)

# 3. Residual Plot (Ambience vs Offers)
X_resid = sm.add_constant(df['Ambience'])
y_resid = df['Offers']

# Fit regression model
model = sm.OLS(y_resid, X_resid).fit()
residuals = model.resid

# Plot residuals
plt.figure(figsize=(8,6))
sns.residplot(x=df['Ambience'], y=y_resid, lowess=True, color="blue")
plt.title("Residual Plot: Ambience vs Offers")
plt.xlabel("Ambience")
plt.ylabel("Residuals")
plt.grid(True)
plt.show()
```
