```python
#LAB Experiment Example 1:  What is the probability of 2 heads in 6 coin tosses? Fit 
#the Binomial distribution for the coin tosses and plot it. 

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom

# Parameters
n = 6
p = 0.5
k = 2

# Probability of exactly 2 heads
prob = binom.pmf(k, n, p)
print("P(X=2):", prob)

# Fit Binomial distribution
x = np.arange(0, n+1)
y = binom.pmf(x, n, p)

# Plot
plt.bar(x, y, color='skyblue', edgecolor='black')
plt.title("Binomial Distribution: 6 Tosses, p=0.5")
plt.xlabel("Number of Heads")
plt.ylabel("Probability")
plt.show()

```

```python

#LAB Experiment Example 2:  A large urban hospital has, on average, 80 emergency 
#department admits every Monday. What is the probability that there will be more than 
#100? give the python code for data and plot the distributio

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import poisson

# Parameters
lam = 80  # average admits
k = 100   # threshold

# Probability of more than 100 admits
prob = 1 - poisson.cdf(k, lam)
print("P(X > 100):", prob)

# Fit Poisson distribution
x = np.arange(50, 120)  # range around mean
y = poisson.pmf(x, lam)

# Plot
plt.bar(x, y, color='skyblue', edgecolor='black')
plt.axvline(k, color='red', linestyle='--', label='Threshold = 100')
plt.title("Poisson Distribution: Emergency Admits (λ=80)")
plt.xlabel("Number of Admits")
plt.ylabel("Probability")
plt.legend()
plt.show()
```
