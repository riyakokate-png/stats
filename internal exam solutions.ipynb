```python
import numpy as np
from scipy.stats import chi2

# Observed and Expected data
observed = np.array([8, 6, 10, 7, 8, 11, 9])
expected = np.array([9, 8, 11, 8, 10, 7, 6])

# Chi-Square calculation
chi_square = np.sum((observed - expected)**2 / expected)

# Degrees of freedom
df = len(observed) - 1

# Critical value at 5% significance level
critical_value = chi2.ppf(0.95, df)

# Display results
print("Chi-Square Statistic:", round(chi_square, 3))
print("Degrees of Freedom:", df)
print("Critical Value at 5% LOS:", round(critical_value, 3))

# Decision
if chi_square < critical_value:
    print("Conclusion: Good fit (Fail to reject H0)")
else:
    print("Conclusion: Not a good fit (Reject H0)")
```
